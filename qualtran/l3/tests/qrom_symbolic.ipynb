### Qubit count

 - base: $m = 0$ gives $n + 1$
 - during the circuit: $2 + L(m)$

### Gate count

 - base: one `C[U]`
 - total: one `And`, one `CNOT`, two $L(m-1)$

```python
L = sympy.Function('L')
rec = L(m) - (2*L(m-1) + 1)
init = {L(0): 0}
sympy.rsolve(rec, L(m), init)
```

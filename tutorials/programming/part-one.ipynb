### Estimate the quantum runtime for an 8-bit adder

```python
from qualtran import QUInt
from qualtran.bloqs.arithmetic import Add
```

```python
op = Add(a_dtype=QUInt(8), b_dtype=QUInt(8))
op
```

```python
from qualtran.resource_counting import get_cost_value, QECGatesCost
```

```python
gc = get_cost_value(op, QECGatesCost())
gc
```

```python
from qualtran.surface_code import PhysicalCostModel, AlgorithmSummary
```

```python
# beyond quadatic
# We have to make a guestimate on a "firmware" parameter.
# following [beyond quadractic, pg. 3 sec: "Implementing error-corrected quantum primitives"], we'll pick d=30
# TODO: make this a default
cost_model = PhysicalCostModel.make_gidney_fowler(data_d=30)
cost_model
```

```python
time = cost_model.duration_s(AlgorithmSummary.from_bloq(op))
time
```

```python
space = cost_model.n_phys_qubits(AlgorithmSummary.from_bloq(op))
space
```

```python
time * space
```

```python
(time * space) / 1e-9
```

### Phase estimate

```python
from qualtran.drawing import show_bloq
```

```python
from qualtran.bloqs.phase_estimation import TextbookQPE, RectangularWindowState
from qualtran.bloqs.basic_gates import ZPowGate, XGate
```

```python
phest_z = TextbookQPE(XGate(), ctrl_state_prep=RectangularWindowState(1))
show_bloq(phest_z.decompose_bloq())
```

```python
cbloq = phest_z.as_composite_bloq().flatten()
show_bloq(cbloq)
```

```python
from qualtran import BloqBuilder
from qualtran.bloqs.basic_gates import ZeroState

bb = BloqBuilder()
x = bb.add(ZeroState())
c, x = bb.add(phest_z, q=x)
cbloq = bb.finalize(c=c, x=x)
```

```python
show_bloq(cbloq)
```

```python
cbloq.tensor_contract()
```
